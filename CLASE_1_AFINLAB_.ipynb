# COLECCIONES

En python podemos trabajar con variables que almacenan multiples valores. estos tipos de variables se denominan colecciones y son :

* LISTAS : [] Ordenados y mutables
* SETS : {} No ordenados e inmutables. No permite duplicados.
* Tuplas : () Ordenados e mutables
* Diccionarios : {key : variable} permite agregar una variable asociada a una    llave key . son mutables.

```python
SELENE = [16,17,18,11]
print(SELENE)
```

```python
# LISTS
Listas = [1,2,3,4,2]
print("Lista:", Listas, type(Listas))
```

```python
# SET TIENE QUE SER CON LLAVES {}
JEANPI_SET = {1,2,3,4,2,3}
print(JEANPI_SET)
```

```python
# Set
set_ = {1, 2, 3, 2}
print("Set:", set_, type(set_))
```

```python
# TUPLAS ES MUY SIMILAR A LAS LISTAS PERO LO QUE LO DIFERENCIA ES QUE SON ()

TUPLAS = (1,2,3,4,2)
print(TUPLAS)




```

```python
# Tuples
tupla = (1, 2, 3, 2)
print("Tupla:", tupla, type(tupla))
```

#INDEXACIÓN Y COMANDOS BÁSICOS

1. count() : contar cuántas veces aparece un elemento específico dentro de una secuencia

2. range() : rango
3. len() : contar cuántos elementos tiene algo.
4. input () : pedir información al usuario.
5. append () : agregar 1 elemento
6. extend () : añadir y acumular varios elementos
7. sort () : Ordenar listas de menor a mayor y si quieres al inversa - sort(reverse=True)
